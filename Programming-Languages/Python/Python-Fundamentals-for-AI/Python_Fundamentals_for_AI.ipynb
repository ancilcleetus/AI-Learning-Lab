# Python Basics: Variables, Types, and Data Structures

Python is a dynamically typed language, meaning you don't need to declare variable types explicitly. The type is inferred from the value you assign. This flexibility makes Python great for rapid prototyping and experimentation, which is why it's the dominant language in AI and machine learning.

## Variables and Types

Python has several built-in types you'll use constantly:

* **int**: Whole numbers (42, -10, 0)
* **float**: Decimal numbers (3.14, -0.5, 2.0)
* **str**: Text strings ("hello", 'world')
* **bool**: True or False values

Unlike statically typed languages, you can reassign a variable to a different type:

```python
x = 5        # x is an int
x = "hello"  # now x is a string - this is fine in Python!
```

You can check a variable's type using the `type()` function, and convert between types using type constructors like `int()`, `float()`, and `str()`.

## Basic Operations

Basic arithmetic operations work as you'd expect:

* Addition: `+`
* Subtraction: `-`
* Multiplication: `*`
* Division: `/` (always returns float)
* Integer division: `//` (rounds down)
* Modulo: `%` (remainder)
* Exponentiation: `**`

## Data Structures

Python provides powerful built-in data structures that you'll use extensively in machine learning:

**Lists** are ordered, mutable collections:

```python
numbers = [1, 2, 3, 4, 5]
numbers.append(6)  # Add to end
numbers[0] = 10    # Modify elements
```

Lists support slicing, a powerful feature for extracting subsequences:

* `numbers[1:4]` → elements at indices 1, 2, 3 (not 4!)
* `numbers[:3]` → first 3 elements
* `numbers[::2]` → every second element
* `numbers[-1]` → last element

**Tuples** are like lists but immutable (can't be modified):

```python
point = (10, 20)  # Often used for coordinates
x, y = point      # Unpacking
```

**Dictionaries** store key-value pairs:

```python
model_config = {
    'learning_rate': 0.001,
    'batch_size': 32,
    'epochs': 10
}
model_config['learning_rate']  # Access values
```

Use `.get()` for safe access when a key might not exist:

```python
print(model_config.get('dropout', 0.0))  # Returns 0.0 if key missing
model_config['dropout']           # Raises KeyError if missing!
```

In deep learning, you'll frequently:

* Store training data in lists
* Use tuples for shapes: `(batch_size, channels, height, width)`
* Pass configuration as dictionaries: `model = Model(**config)`
* Work with different numeric types (int32, float32, float64) that affect memory usage and computation speed

# Functions and Arguments

Functions are the building blocks of reusable code. In Python, functions are incredibly flexible with multiple ways to pass arguments.

**Basic function syntax**:

```python
def function_name(param1, param2):
    result = param1 + param2
    return result
```

**Default arguments** provide fallback values:

```python
def train_model(epochs=10, lr=0.001):
    # epochs defaults to 10 if not provided
    pass
```

**Keyword arguments** make function calls clearer:

```python
train_model(epochs=20, lr=0.01)  # Explicit and readable
train_model(lr=0.01, epochs=20)  # Order doesn't matter!
```

***args** accepts any number of positional arguments:

```python
def sum_all(*numbers):
    return sum(numbers)

sum_all(1, 2, 3, 4, 5)  # Works with any number of args
```

****kwargs** accepts any number of keyword arguments:

```python
def create_model(**config):
    # config is a dictionary of all keyword args
    lr = config.get('learning_rate', 0.001)
    return Model(lr)

create_model(learning_rate=0.01, batch_size=32)
```

In PyTorch and deep learning libraries, you'll see patterns like:

```python
model = Model(hidden_size=128, num_layers=3, dropout=0.1)
optimizer = Adam(model.parameters(), lr=0.001)
```

These use keyword arguments extensively for clarity and flexibility. Understanding how to define and call functions with different argument types is essential for working with ML libraries.

# Control Flow and Comprehensions

Control flow structures let you make decisions and repeat operations. Python's control flow is clean and powerful.

**Conditional statements**:

```python
if loss < 0.01:
    print("Model converged!")
elif loss < 0.1:
    print("Getting close...")
else:
    print("Keep training")
```

**For loops** iterate over sequences:

```python
for epoch in range(10):  # 0 through 9
    train_model()

for item in data_list:
    process(item)
```

Useful iteration functions:

* `range(10)` → 0, 1, 2, ..., 9
* `enumerate(list)` → (0, item0), (1, item1), ...
* `zip(list1, list2)` → pairs of elements

**List comprehensions** create lists concisely:

```python
squares = []
for x in range(10):  # Traditional loop
    squares.append(x**2)

squares = [x**2 for x in range(10)]  # List comprehension (Pythonic!)

evens = [x for x in range(20) if x % 2 == 0]  # With condition
```

**Dictionary comprehensions** work similarly:

```python
weights = {f'layer_{i}': torch.randn(64, 64) for i in range(5)}
```

In deep learning code, you'll frequently see:

```python
for batch_idx, (data, target) in enumerate(train_loader):  # Process each batch
    # Training code
    pass

activations = [layer(x) for layer in model.layers]  # Create list of layer outputs
```

Comprehensions make code more readable and often faster than equivalent loops.

# Classes and `__init__`

Classes are templates for creating objects. They bundle data (attributes) and behavior (methods) together. In deep learning, almost everything is a class: models, layers, optimizers, datasets.

**Basic class syntax**:

```python
class Dog:
    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age
    
    def bark(self):
        return f"{self.name} says woof!"

my_dog = Dog("Buddy", 3)  # Create instance
print(my_dog.bark())  # "Buddy says woof!"
```

The **`__init__` method** is the constructor - it runs when you create a new instance. It initializes the object's state.

The **`self` parameter** refers to the instance being created. It's always the first parameter of instance methods.

**Instance attributes** (`self.name`, `self.age`) store data specific to each instance.

**Instance methods** are functions defined inside a class that operate on instances.

In PyTorch, you'll see this pattern constantly:

```python
class NeuralNetwork(nn.Module):
    def __init__(self, input_size, hidden_size):
        super().__init__()
        self.layer1 = nn.Linear(input_size, hidden_size)
        self.layer2 = nn.Linear(hidden_size, 10)
    
    def forward(self, x):
        x = self.layer1(x)
        x = torch.relu(x)
        x = self.layer2(x)
        return x

model = NeuralNetwork(784, 128)
output = model(input_data)  # Calls forward method
```

Understanding classes is fundamental because neural networks, layers, datasets, and optimizers are all implemented as classes in PyTorch.
