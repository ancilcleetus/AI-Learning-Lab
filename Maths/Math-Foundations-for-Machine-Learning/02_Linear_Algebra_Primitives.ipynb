# Vector Element-wise Sum

**Problem**

Write a Python function that computes the element-wise sum of two vectors. The function should return a new vector representing the resulting sum if the operation is valid, or -1 if the vectors have incompatible dimensions. Two vectors (lists) can be summed element-wise only if they are of the same length.

```python
import numpy as np

def vector_element_wise_sum(a: list[int | float], b: list[int | float]) -> list[int | float]:
    """
    Return the element wise sum of vectors a & b
    Return -1 if the vectors have different lengths
    """
    if len(a) != len(b):
        return -1

    c = np.array(a) + np.array(b)
    return c.tolist()
```

```python
print(vector_element_wise_sum([1, 2, 3], [4, 5, 6]))
```

```python
import torch

def vector_element_wise_sum(a: torch.Tensor, b: torch.Tensor) -> torch.Tensor:
    """
    Return the element wise sum of vectors a & b
    Return -1 if the vectors have different lengths
    """
    if a.shape != b.shape:
        return -1

    c = a + b
    return c
```

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
c = vector_element_wise_sum(a, b)
print(c)
d = c.numpy()
print(d)
e = d.tolist()
print(e)
```

# Dot Product Calculator

**Problem**

Write a Python function to calculate the dot product of two vectors. The function should take two 1D NumPy arrays as input and return the dot product as a single number.

You may assume both vectors have the same length.

```python
import numpy as np

def calculate_dot_product(a: np.ndarray, b: np.ndarray) -> float:
    """
    Return the dot product of vectors a & b
    """
    # return np.dot(a, b)
    # return a.dot(b)
    # return np.sum(a * b)
    return (a * b).sum()
```

```python
print(calculate_dot_product(np.array([1, 2, 3]), np.array([4, 5, 6])))
```

```python
import torch

def calculate_dot_product(a: torch.Tensor, b: torch.Tensor) -> torch.Tensor:
    """
    Return the dot product of vectors a & b
    """
    # return torch.dot(a, b)
    # return a.dot(b)
    # return torch.sum(a * b)
    return (a * b).sum()
```

```python
dot_product = calculate_dot_product(torch.tensor([1, 2, 3]), torch.tensor([4, 5, 6]))
print(dot_product)
print(f"Dot Product: {dot_product.item()}")
```

# Vector Norms (L1/L2/Frobenius)

**Problem**

Implement a function that computes different types of norms for vectors and matrices.

Vector norms are fundamental concepts in linear algebra that measure the "size" or "length" of vectors. Different norms have different properties and applications in machine learning, particularly in regularization and distance calculations.

```python
import numpy as np

def compute_norm(arr: np.ndarray, norm_type: str) -> float:
    """
    Return the norm of the vector or matrix
    """
    if norm_type == "L1":
        # return np.linalg.norm(arr, ord=1)
        return np.sum(np.abs(arr), axis=0)
    elif norm_type == "L2":
        # return np.linalg.norm(arr, ord=2)
        return np.sqrt(np.sum(np.square(arr), axis=0))
    elif norm_type == "Frobenius":
        if arr.ndim == 2:
            # return np.linalg.norm(arr, ord="fro")
            return np.sqrt(np.sum(np.square(arr)))
        else:
            return "Invalid input: input must be a 2D array"
    else:
        return "Invalid norm type"
```

```python
print(compute_norm(np.array([1, -2, 3]), "L1"))
print(compute_norm(np.array([3, 4]), "L2"))
print(compute_norm(np.array([1.5, -2.5, 3.0]), "L1"))
print(compute_norm(np.array([1, 2, 2]), "L2"))
print(compute_norm(np.array([[1, 2], [3, 4]]), "Frobenius"))
```

```python
import torch

def compute_norm(arr: torch.Tensor, norm_type: str) -> float:
    """
    Return the norm of the vector or matrix
    """
    arr = arr.float()
    if norm_type == "L1":
        # return torch.norm(arr, p=1).item()
        return torch.sum(torch.abs(arr), axis=0).item()
    elif norm_type == "L2":
        # return torch.norm(arr, p=2).item()
        return torch.sqrt(torch.sum(torch.square(arr), axis=0)).item()
    elif norm_type == "Frobenius":
        if arr.ndim == 2:
            # return torch.norm(arr, p="fro").item()
            return torch.sqrt(torch.sum(torch.square(arr))).item()
        else:
            return "Invalid input: input must be a 2D tensor"
    else:
        return "Invalid norm type"
```

```python
print(compute_norm(torch.tensor([1, -2, 3]), "L1"))
print(compute_norm(torch.tensor([3, 4]), "L2"))
print(compute_norm(torch.tensor([1.5, -2.5, 3.0]), "L1"))
print(compute_norm(torch.tensor([1, 2, 2]), "L2"))
print(compute_norm(torch.tensor([[1, 2], [3, 4]]), "Frobenius"))
```

# Calculate Cosine Similarity between Vectors

**Problem**

In this task, you need to implement a function cosine_similarity(v1, v2) that calculates the cosine similarity between two vectors. Cosine similarity measures the cosine of the angle between two vectors, indicating their directional similarity.

```python
import numpy as np

def cosine_similarity(v1: np.ndarray, v2: np.ndarray) -> float:
    """
    Return the cosine similarity (cosine of angle betweenn two vectors)
    """
    # dot_product = np.dot(v1, v2)
    # v1_l2 = np.linalg.norm(v1, ord=2)
    # v2_l2 = np.linalg.norm(v2, ord=2)

    dot_product = np.sum(v1 * v2)
    v1_l2 = np.sqrt(np.sum(np.square(v1)))
    v2_l2 = np.sqrt(np.sum(np.square(v2)))
    return dot_product / (v1_l2 * v2_l2)
```

```python
print(cosine_similarity(np.array([1, 2, 3]), np.array([2, 4, 6])))
print(cosine_similarity(np.array([1, 2, 3]), np.array([-1, -2, -3])))
print(cosine_similarity(np.array([1, 2, 3]), np.array([0, 4, -1])))
```

```python
import torch

def cosine_similarity(v1: torch.Tensor, v2: torch.Tensor) -> float:
    """
    Return the cosine similarity (cosine of angle betweenn two vectors)
    """
    v1 = v1.float()
    v2 = v2.float()

    # dot_product = torch.dot(v1, v2)
    # v1_l2 = torch.norm(v1, p=2)
    # v2_l2 = torch.norm(v2, p=2)
    # similarity = dot_product / (v1_l2 * v2_l2)

    dot_product = torch.sum(v1 * v2)
    v1_l2 = torch.sqrt(torch.sum(torch.square(v1)))
    v2_l2 = torch.sqrt(torch.sum(torch.square(v2)))
    similarity = dot_product / (v1_l2 * v2_l2)

    return similarity.item()
```

```python
print(cosine_similarity(torch.tensor([1, 2, 3]), torch.tensor([2, 4, 6])))
print(cosine_similarity(torch.tensor([1, 2, 3]), torch.tensor([-1, -2, -3])))
print(cosine_similarity(torch.tensor([1, 2, 3]), torch.tensor([0, 4, -1])))
```
