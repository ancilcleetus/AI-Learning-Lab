# Vector Element-wise Sum

## Understanding Vector Element-wise Sum

In linear algebra, the **element-wise sum** (also known as vector addition) involves adding corresponding entries of two vectors.

**Vector Notation:**

Given two vectors $a$ and $b$ of the same dimension $n$:
\begin{equation*}
a = \begin{pmatrix}
a_1 \\
a_2 \\
\vdots \\
a_n
\end{pmatrix}, \quad b = \begin{pmatrix}
b_1 \\
b_2 \\
\vdots \\
b_n
\end{pmatrix}
\end{equation*}

The element-wise sum is defined as:
\begin{equation*}
a + b = \begin{pmatrix}
a_1 + b_1 \\
a_2 + b_2 \\
\vdots \\
a_n + b_n
\end{pmatrix}
\end{equation*}

**Key Requirement:**

Vectors $a$ and $b$ must be of the same length $n$ for the operation to be valid. If their lengths differ, element-wise addition is not defined.

**Example:**

Let:
\begin{equation*}
a = [1, 2, 3], \quad b = [4, 5, 6]
\end{equation*}

Then:
\begin{equation*}
a + b = [1 + 4, 2 + 5, 3 + 6] = [5, 7, 9]
\end{equation*}

This simple operation is foundational in many applications such as vector arithmetic, neural network computations, and linear transformations.

## Problem: Function that computes the element-wise sum of two vectors

**Problem:**

Write a Python function that computes the element-wise sum of two vectors. The function should return a new vector representing the resulting sum if the operation is valid, or -1 if the vectors have incompatible dimensions. Two vectors (lists) can be summed element-wise only if they are of the same length.

### NumPy Solution

```python
import numpy as np

def vector_element_wise_sum(a: list[int | float], b: list[int | float]) -> list[int | float]:
    """
    Return the element wise sum of vectors a & b
    Return -1 if the vectors have different lengths
    """
    if len(a) != len(b):
        return -1

    c = np.array(a) + np.array(b)
    return c.tolist()
```

```python
print(vector_element_wise_sum([1, 2, 3], [4, 5, 6]))
```

### PyTorch Solution

```python
import torch

def vector_element_wise_sum(a: torch.Tensor, b: torch.Tensor) -> torch.Tensor:
    """
    Return the element wise sum of vectors a & b
    Return -1 if the vectors have different lengths
    """
    if a.shape != b.shape:
        return -1

    c = a + b
    return c
```

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([4, 5, 6])
c = vector_element_wise_sum(a, b)
print(c)
d = c.numpy()
print(d)
e = d.tolist()
print(e)
```

# Dot Product Calculator

## Calculating the Dot Product of Two Vectors

The dot product, also known as the scalar product, is a mathematical operation that takes two equal-length vectors and returns a single number. It is widely used in physics, geometry, and linear algebra.

**Formula for the Dot Product:**

The dot product of two vectors $a$ and $b$, each of length $n$, is calculated as follows:
\begin{equation*}
a \cdot b = \sum_{i=1}^{n} a_i b_i
\end{equation*}

This means multiplying corresponding elements of the two vectors and summing up the results.

**Geometric Interpretation:**

In geometric terms, the dot product can also be expressed as:
\begin{equation*}
a \cdot b = \left| a \right| \left| b \right| \cos \theta
\end{equation*}

where:
* $\left| a \right|$ and $\left| b \right|$ are the magnitudes of the vectors
* $\theta$ is the angle between the two vectors

**Properties of the Dot Product:**

1. Commutative:
\begin{equation*}
a \cdot b = b \cdot a
\end{equation*}

2. Distributive:
\begin{equation*}
a \cdot (b+c) = a \cdot b + a \cdot c
\end{equation*}

3. Orthogonal Vectors:

    If $a \cdot b = 0$, then $a$ and $b$ are perpendicular.

**Example Calculation:**

Given two vectors:
* $a = [1, 2, 3]$
* $b = [4, 5, 6]$

the dot product is calculated as:
\begin{equation*}
a \cdot b = (1 \cdot 4) + (2 \cdot 5) + (3 \cdot 6) = 4 + 10 + 18 = 32
\end{equation*}

**Conclusion:**

The dot product is a fundamental operation in vector algebra, useful in determining angles between vectors, projections, and in many applications across physics and engineering.

# Vector Norms (L1/L2/Frobenius)

## Vector Norms in Linear Algebra

Norms are functions that assign a non-negative length or size to vectors. They are essential in machine learning for measuring distances, implementing regularization, and analyzing convergence.

**L1 Norm (Manhattan Norm):**

The L1 norm, also called the Manhattan or taxicab norm, is the sum of the absolute values of all elements:
\begin{equation*}
\left\| X \right\| _1 = \sum_{i=1}^{n} \left| x_i \right|
\end{equation*}

For a vector $X = [x_1, x_2,...,x_n]$, this measures the total distance traveled along axis-aligned paths.

Applications:

* L1 regularization (Lasso regression) promotes sparsity (many weights reduced to zero) $\Longrightarrow$ Feature selection
* Robust to outliers in certain contexts

**L2 Norm (Euclidean Norm):**

The L2 norm is the most common norm, representing the straight-line distance from the origin:
\begin{equation*}
\left\| X \right\| _2 = \sqrt{\sum_{i=1}^{n} x_i^2}
\end{equation*}

This is the standard Euclidean length of a vector.

Applications:

* L2 regularization (Ridge regression) prevents large weights
* Distance calculations in k-NN and clustering
* Gradient magnitude in optimization

**Frobenius Norm:**

The Frobenius norm extends the L2 norm concept to matrices. For a matrix $A$ with elements $a_{ij}$:
\begin{equation*}
\left\| A \right\| _F = \sqrt{\sum_{i=1}^{m} \sum_{j=1}^{n} a_{ij}^2}
\end{equation*}

This is equivalent to treating the matrix as a flattened vector and computing its L2 norm.

Applications:

* Measuring matrix approximation error
* Nuclear norm approximation in matrix factorization
* Weight matrix analysis in neural networks

**Properties of Norms:**

All norms satisfy these properties for vectors $x$, $y$ and scalar $\alpha$:
1. Non-negativity: $\left\| X \right\| \ge 0$, with equality iff $X = 0$
2. Scalar multiplication: $\left\| \alpha X \right\| = \left| \alpha \right| \left\| X \right\|$
3. Triangle inequality: $\left\| X + Y \right\| \le \left\| X \right\| + \left\| Y \right\|$

**Relationship Between Norms:**

For any vector, the norms satisfy $\left\| X \right\| _2 \le \left\| X \right\| _1 \le \sqrt{n} \left\| X \right\| _2$ where $n$ is the dimension.

## Problem: Function that computes different types of norms for vectors and matrices

**Problem:**

Implement a function that computes different types of norms for vectors and matrices.

Vector norms are fundamental concepts in linear algebra that measure the "size" or "length" of vectors. Different norms have different properties and applications in machine learning, particularly in regularization and distance calculations.

**Task:**

Write a function `compute_norm(arr, norm_type)` that takes:

* `arr`: A numpy array (can be 1D or 2D)
* `norm_type`: A string specifying which norm to compute (`'L1'`, `'L2'`, or `'Frobenius'`)

The function should return the computed norm as a float.

**Constraints:**

* For `'L1'` and `'L2'` norms, the input can be any array (1D or 2D)
* For `'Frobenius'` norm, the input is typically a 2D matrix
* All norms should treat the input as a flattened collection of elements when computing

### NumPy Solution

```python
import numpy as np

def compute_norm(arr: np.ndarray, norm_type: str) -> float:
    """
    Return the norm of the vector or matrix
    """
    # Solution 1
    if norm_type == "L1":
        return np.linalg.norm(arr, ord=1)
    elif norm_type == "L2":
        return np.linalg.norm(arr, ord=2)
    elif norm_type == "Frobenius":
        if arr.ndim == 2:
            return np.linalg.norm(arr, ord="fro")
        else:
            return "Invalid input: input must be a 2D array"
    else:
        return "Invalid norm type"

    # Solution 2
    if norm_type == "L1":
        return np.sum(np.abs(arr), axis=0)
    elif norm_type == "L2":
        return np.sqrt(np.sum(np.square(arr), axis=0))
    elif norm_type == "Frobenius":
        if arr.ndim == 2:
            return np.sqrt(np.sum(np.square(arr)))
        else:
            return "Invalid input: input must be a 2D array"
    else:
        return "Invalid norm type"
```

```python
print(compute_norm(np.array([1, -2, 3]), "L1"))
print(compute_norm(np.array([3, 4]), "L2"))
print(compute_norm(np.array([1.5, -2.5, 3.0]), "L1"))
print(compute_norm(np.array([1, 2, 2]), "L2"))
print(compute_norm(np.array([[1, 2], [3, 4]]), "Frobenius"))
```

### PyTorch Solution

```python
import torch

def compute_norm(arr: torch.Tensor, norm_type: str) -> float:
    """
    Return the norm of the vector or matrix
    """
    # Solution 1
    arr = arr.float()
    if norm_type == "L1":
        return torch.norm(arr, p=1).item()
    elif norm_type == "L2":
        return torch.norm(arr, p=2).item()
    elif norm_type == "Frobenius":
        if arr.ndim == 2:
            return torch.norm(arr, p="fro").item()
        else:
            return "Invalid input: input must be a 2D tensor"
    else:
        return "Invalid norm type"

    # Solution 2
    arr = arr.float()
    if norm_type == "L1":
        return torch.sum(torch.abs(arr), axis=0).item()
    elif norm_type == "L2":
        return torch.sqrt(torch.sum(torch.square(arr), axis=0)).item()
    elif norm_type == "Frobenius":
        if arr.ndim == 2:
            return torch.sqrt(torch.sum(torch.square(arr))).item()
        else:
            return "Invalid input: input must be a 2D tensor"
    else:
        return "Invalid norm type"
```

```python
print(compute_norm(torch.tensor([1, -2, 3]), "L1"))
print(compute_norm(torch.tensor([3, 4]), "L2"))
print(compute_norm(torch.tensor([1.5, -2.5, 3.0]), "L1"))
print(compute_norm(torch.tensor([1, 2, 2]), "L2"))
print(compute_norm(torch.tensor([[1, 2], [3, 4]]), "Frobenius"))
```

# Calculate Cosine Similarity between Vectors

## Cosine Similarity

Cosine similarity measures the cosine of the angle between two vectors. It doesn't consider the magnitude of the vectors but focuses on the angle between them.

**Cosine Similarity Formula:**

\begin{equation*}
\cos(\theta) = \frac{\sum_{i=1}^{p} A_i B_i}{\sqrt{\sum_{i=1}^{p} A_i^2}\sqrt{\sum_{i=1}^{p} B_i^2}}
\end{equation*}

**Implementation Steps for Cosine Similarity:**

1. Handle Input: Ensure input vectors have the same dimensions and handle edge cases (e.g., zero vectors)
2. Dot Product: Compute $\sum_{i=1}^{p} A_i B_i$ for the two vectors
3. Magnitudes: Compute the L2 norms $\sqrt{\sum_{i=1}^{p} A_i^2}$ and $\sqrt{\sum_{i=1}^{p} B_i^2}$
4. Final Result: Divide the dot product by the product of the magnitudes.

**Use Cases:**

* Text and Image Similarity
* Recommendation Systems
* Query Matching

**Pitfalls:**

1. Magnitude Blindness:
    * Example:
        * $\text{vector1} = (1, 1)$
        * $\text{vector2} = (1000, 1000)$
        * Cosine similarity = 1, despite the vastly different magnitudes
2. Sparse Data Issues:
    * In high-dimensional spaces, where data is often sparse, cosine similarity may become less reliable.
3. Non-Negative Data Limitation:
    * If all values are positive, cosine similarity cannot capture negative relationships or inverse trends.

# Transpose of a Matrix

## Understanding Transpose of a Matrix

The **transpose** of a matrix is obtained by interchanging its rows and columns. If $A$ is an $m \times n$ matrix, then its transpose $A^T$ is an $n \times m$ matrix.

**Definition:**

For a matrix $A$ where element $A_{ij}$ is in row $i$ and column $j$:
\begin{equation*}
A^T_{ij} = A_{ji}
\end{equation*}

**Example:**

Original Matrix $A$ (2×3):
\begin{equation*}
A = \begin{pmatrix}
1 & 2 & 3 \\
4 & 5 & 6
\end{pmatrix}
\end{equation*}

Transposed Matrix $A$ (3×2):
\begin{equation*}
A^T = \begin{pmatrix}
1 & 4 \\
2 & 5 \\
3 & 6
\end{pmatrix}
\end{equation*}

**Properties of Transpose:**

1. Double transpose: $(A^T)^T = A$
2. Sum: $(A + B)^T = A^T + B^T$
3. Scalar multiplication: $(cA)^T = cA^T$
4. Product: $(AB)^T = B^T A^T$ (note the reversed order)
5. Symmetric matrix: $A = A^T$ means $A$ is symmetric

**Python Implementation:**

Using `zip(*matrix)` is a Pythonic way to transpose:

* `*matrix` unpacks rows as separate arguments
* `zip()` groups elements by position (first elements together, second elements together, etc.)
* Convert each tuple to a list for the final result

**Applications:**

* Linear algebra: Solving systems of equations
* Machine learning: Reshaping feature matrices, computing covariance matrices
* Computer graphics: Transformation matrices
* Data processing: Converting between row-major and column-major formats

## Problem: Function that computes the transpose of a given 2D matrix

**Problem:**

Write a Python function that computes the transpose of a given 2D matrix. The transpose of a matrix is formed by turning its rows into columns and columns into rows. For an m×n matrix, the transpose will be an n×m matrix.

**Example:**
* Input: `a = [[1, 2, 3], [4, 5, 6]]`
* Output: `[[1, 4], [2, 5], [3, 6]]`
* Reasoning:

    The input is a 2×3 matrix. The transpose swaps rows and columns: the first row [1, 2, 3] becomes the first column, and the second row [4, 5, 6] becomes the second column, resulting in a 3×2 matrix.


### NumPy Solution

```python
import numpy as np

def get_transpose_matrix(a: list[list[int | float]]) -> list[list[int | float]]:
    """
    Transpose a 2D matrix by swapping rows and columns.

    Args:
        a: A 2D matrix of shape (m, n)

    Returns:
        The transposed matrix of shape (n, m)
    """
    # Solution 1
    # matrix = np.array(a)
    # transposed_matrix = matrix.T
    # return transposed_matrix.tolist()

    # Solution 2
    # matrix = np.array(a)
    # transposed_matrix = np.transpose(matrix)
    # return transposed_matrix.tolist()

    # Solution 3: zip(*a) returns transposed tuples
    transposed_matrix = [list(row) for row in zip(*a)]
    return transposed_matrix
```

```python
print(get_transpose_matrix([[1, 2], [3, 4], [5, 6]]))
print(get_transpose_matrix([[1, 2, 3], [4, 5, 6]]))
print(get_transpose_matrix([[1, 2], [3, 4]]))  # Square matrix
print(get_transpose_matrix([[1, 2, 3]]))  # Row vector to column vector
print(get_transpose_matrix([[1], [2], [3]]))  # Column vector to row vector
```

### PyTorch Solution

```python
import torch

def get_transpose_matrix(a) -> list[list[int | float]]:
    """
    Transpose a 2D matrix using PyTorch.

    Args:
        a: A 2D matrix (can be list, numpy array, or torch.Tensor)

    Returns:
        A transposed torch.Tensor
    """
    # Solution 1
    # matrix = torch.as_tensor(a)
    # transposed_matrix = matrix.T
    # return transposed_matrix.numpy().tolist()

    # Solution 2
    # matrix = torch.as_tensor(a)
    # transposed_matrix = torch.transpose(matrix, 0, 1)  # For a 2D matrix, you swap dimension 0 (rows) and dimension 1 (columns)
    # return transposed_matrix.numpy().tolist()

    # Solution 3
    matrix = torch.as_tensor(a)
    transposed_matrix = torch.permute(matrix, (1, 0))  # We pass the matrix first, then the new order of dimensions (1, 0)
    return transposed_matrix.numpy().tolist()
```

# Scalar Multiplication of a Matrix

## Understanding Scalar Multiplication of a Matrix

When a matrix $A$ is multiplied by a scalar $k$, the operation is defined as multiplying each element of $A$ by $k$.

Given a matrix $A$:
\begin{equation*}
A = \begin{pmatrix}
a_{11} & a_{12} \\
a_{21} & a_{22}
\end{pmatrix}
\end{equation*}

and a scalar $k$, the result of the scalar multiplication $kA$ is:
\begin{equation*}
kA = \begin{pmatrix}
ka_{11} & ka_{12} \\
ka_{21} & ka_{22}
\end{pmatrix}
\end{equation*}

This operation scales the matrix by $k$ without changing its dimension or the relative proportion of its elements.

## Problem: Function that multiplies a matrix by a scalar and returns the result

**Problem:**

Write a Python function that multiplies a matrix by a scalar and returns the result.

**Example:**
* Input: `matrix = [[1, 2], [3, 4]], scalar = 2`
* Output: `[[2, 4], [6, 8]`

**Reasoning:**

Each element of the matrix is multiplied by the scalar.

### NumPy Solution

```python
import numpy as np

def scalar_multiply(matrix: list[list[int | float]], scalar: int | float) -> list[list[int | float]]:
    matrix = np.array(matrix)
    scaled_matrix = matrix * scalar
    return scaled_matrix.tolist()
```

```python
print(scalar_multiply([[1,2],[3,4]], 2))
print(scalar_multiply([[0,-1],[1,0]], -1))
print(scalar_multiply([[1]], 0))
print(scalar_multiply([[1.5, -2.0], [0.0, 3.5]], 2))
```

### PyTorch Solution

```python
import torch

def scalar_multiply(matrix, scalar) -> torch.Tensor:
    """
    Multiply each element of a 2D matrix by a scalar using PyTorch.
    Inputs can be Python lists, NumPy arrays, or torch Tensors.
    Returns a 2D tensor of the same shape.
    """
    # Convert input to tensor
    matrix = torch.as_tensor(matrix)
    scaled_matrix = matrix * scalar
    return scaled_matrix
```

```python
print(scalar_multiply([[1,2],[3,4]], 2).numpy().tolist())
print(scalar_multiply([[0,-1],[1,0]], -1).numpy().tolist())
print(scalar_multiply([[1]], 0).numpy().tolist())
print(scalar_multiply([[1.5, -2.0], [0.0, 3.5]], 2).numpy().tolist())
```

# Matrix-Vector Dot Product

## Understanding Matrix-Vector Dot Product

Consider a matrix $A$ and a vector $v$.

Matrix $A$ (n x m):
\begin{equation*}
A = \begin{pmatrix}
a_{11} & a_{12} & \cdots & a_{1m} \\
a_{21} & a_{22} & \cdots & a_{2m} \\
\cdot & \cdot & \cdots & \cdot \\
\cdot & \cdot & \cdots & \cdot \\
\cdot & \cdot & \cdots & \cdot \\
a_{n1} & a_{n2} & \cdots & a_{nm}
\end{pmatrix}
\end{equation*}

Vector $v$ (length m):
\begin{equation*}
v = \begin{pmatrix}
v_1 \\
v_2 \\
\cdot \\
\cdot \\
\cdot \\
v_m
\end{pmatrix}
\end{equation*}

The dot product $A \cdot v$ produces a new vector of length $n$:
\begin{equation*}
A \cdot v = \begin{pmatrix}
a_{11}v_1 + a_{12}v_2 + \cdots + a_{1m}v_m \\
a_{21}v_1 + a_{22}v_2 + \cdots + a_{2m}v_m \\
\cdot \\
\cdot \\
\cdot \\
a_{n1}v_1 + a_{n2}v_2 + \cdots + a_{nm}v_m
\end{pmatrix}
\end{equation*}

**Key Requirement:**

The number of columns in the matrix ($m$) must equal the length of the vector ($m$). If not, the operation is undefined, and the function should return -1.

## Problem: Function that computes the dot product of a matrix and a vector

**Problem:**

Write a Python function that computes the dot product of a matrix and a vector. The function should return a list representing the resulting vector if the operation is valid, or -1 if the matrix and vector dimensions are incompatible. A matrix (a list of lists) can be dotted with a vector (a list) only if the number of columns in the matrix equals the length of the vector. For example, an n x m matrix requires a vector of length m.

**Example:**
* Input: `a = [[1, 2], [2, 4]], b = [1, 2]`
* Output: `[5, 10]`

**Reasoning:**

Dot product of a matrix and a vector $\Longrightarrow$ row by row dot product with the vector

Row 1: $(1 * 1) + (2 * 2) = 1 + 4 = 5$

Row 2: $(2 * 1) + (4 * 2) = 2 + 8 = 10$

### NumPy Solution

```python
import numpy as np

def matrix_dot_vector(matrix: list[list[int | float]], vector: list[int | float]) -> list[int | float]:
    """
    Return a list where each element is the dot product of a row of 'matrix' with 'vector'.
    If the number of columns in 'matrix' does not match the length of 'vector', return -1.
    """
    matrix, vector = np.array(matrix), np.array(vector)
    if matrix.shape[-1] != vector.shape[0]:
        return -1

    # Solution 1
    # return (matrix.dot(vector)).tolist()

    # Solution 2
    # return (matrix @ vector).tolist()

    # Solution 3
    # return np.matmul(matrix, vector).tolist()

    # Solution 4
    return (matrix * vector).sum(axis=1).tolist()
```

```python
print(matrix_dot_vector([[1, 2, 3], [2, 4, 5], [6, 8, 9]], [1, 2, 3]))
print(matrix_dot_vector([[1, 2], [2, 4], [6, 8], [12, 4]], [1, 2, 3]))
print(matrix_dot_vector([[1.5, 2.5], [3.0, 4.0]], [2, 1]))
```

### PyTorch Solution

```python
import torch

def matrix_dot_vector(matrix, vector) -> torch.Tensor:
    """
    Compute the product of matrix `matrix` and vector `vector` using PyTorch.
    Inputs can be Python lists, NumPy arrays, or torch Tensors.
    Returns a 1-D tensor of length m, or tensor(-1) if dimensions mismatch.
    """
    matrix = torch.as_tensor(matrix, dtype=torch.float)
    vector = torch.as_tensor(vector, dtype=torch.float)
    if matrix.shape[-1] != vector.shape[0]:
        return torch.tensor(-1)

    # Solution 1
    # return torch.matmul(matrix, vector)

    # Solution 2
    return matrix @ vector
```

```python
print(matrix_dot_vector([[1, 2, 3], [2, 4, 5], [6, 8, 9]], [1, 2, 3]).numpy().tolist())
print(matrix_dot_vector([[1, 2], [2, 4], [6, 8], [12, 4]], [1, 2, 3]).numpy().tolist())
print(matrix_dot_vector([[1.5, 2.5], [3.0, 4.0]], [2, 1]).numpy().tolist())
```
