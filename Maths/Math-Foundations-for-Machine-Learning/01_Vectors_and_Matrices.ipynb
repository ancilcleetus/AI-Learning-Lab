# Vector Operations

* Vectors add element-wise: (a, b) + (c, d) = (a+c, b+d)
* Scalar multiplication scales each component: k*(a, b) = (ka, kb)
* The dot product is u · v = u1v1 + u2v2 + ... + un*vn — a scalar that measures alignment between vectors

# Matrix Basics

* An m×n matrix has m rows and n columns.
* The transpose A^T flips rows and columns: (A^T){ij} = A{ji}.
* The identity matrix I has 1s on the diagonal and 0s elsewhere; it satisfies AI = IA = A for any appropriately-sized A.

![rainbow](https://raw.githubusercontent.com/ancilcleetus/AI-Learning-Lab/main/assets/rainbow-divider.png)

# Matrix Multiplication

* If A is m×k and B is k×n, then C = AB is m×n with C_{ij} = sum over l of A_{il} * B_{lj}.
* The inner dimensions must match.
* Matrix multiplication is:
    * associative (AB)C = A(BC)
    * distributive A(B+C) = AB + AC
    * but generally NOT commutative: AB ≠ BA in general.

![rainbow](https://raw.githubusercontent.com/ancilcleetus/AI-Learning-Lab/main/assets/rainbow-divider.png)

# Vector Norms and Linear Independence

* Common norms:
    * L1 (Manhattan) ||v||_1 = sum of absolute values
    * L2 (Euclidean) ||v||2 = sqrt(sum of squares)
    * L∞ ||v||∞ = max absolute value
* Vectors are linearly independent if no non-trivial combination c1v1 + c2v2 + ... = 0 — equivalently, none is a combination of the others
    * Trivial solution $\Longrightarrow$ c1 = 0, c2 = 0, ..., cn = 0
