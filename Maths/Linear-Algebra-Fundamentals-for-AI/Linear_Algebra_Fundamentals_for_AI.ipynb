# What is a Vector?

A vector is simply an ordered list of numbers. Think of it as a way to store multiple related values together.

**Everyday examples**:

* Your location on a map: [latitude, longitude] - that's a 2D vector!
* A color: [red, green, blue] like [255, 128, 0] for orange - a 3D vector
* A student's grades: [Math: 85, Science: 92, English: 78] - a 3D vector

**Mathematical notation**:

A vector is written as: v = [v₁, v₂, ..., vₙ]

Examples:

* 2D vector: [3, 4] (two numbers)
* 3D vector: [1, 2, 3] (three numbers)
* You can have vectors with hundreds or thousands of numbers!

**Picture it**:

A 2D vector [3, 4] is like an arrow:

* Starting point: the origin (0,0)
* Ending point: (3, 4)
* Direction: 3 units right, 4 units up
* Length: √(3² + 4²) = 5

**In Python with NumPy**:

```python
import numpy as np
vec = np.array([3, 4])  # Create a vector
length = np.linalg.norm(vec)  # Compute length = 5.0
```

**Why vectors matter for AI**:

Computers can't understand words or images directly. Instead, we convert everything into vectors of numbers. An image becomes a long list of pixel values. A word becomes a list of numbers that capture its meaning. This course will show you how!

# Vector Addition

Vector addition combines two vectors by adding their matching components. It's like adding two shopping lists together!

**How it works**:

For vectors u = [u₁, u₂, ..., uₙ] and v = [v₁, v₂, ..., vₙ]:
u + v = [u₁+v₁, u₂+v₂, ..., uₙ+vₙ]

**Simple example**:

[1, 2] + [3, 4] = [1+3, 2+4] = [4, 6]

Just add the first numbers together, then the second numbers, and so on!

**Real-world example**:

If Monday's sales are [10 apples, 5 oranges] and Tuesday's are [8 apples, 12 oranges]:
Total = [10+8, 5+12] = [18 apples, 17 oranges]

**Picture it (tip-to-tail method)**:

1. Draw the first arrow
2. Start the second arrow where the first one ends
3. The sum is an arrow from the start of the first to the end of the second

**Useful properties**:

* Order doesn't matter: u + v = v + u
* Adding zero changes nothing: v + [0,0] = v

**In Python with NumPy**:

```python
import numpy as np
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
sum_vec = u + v  # [5, 7, 9]
```

NumPy makes this easy - just use the `+` sign! No loops needed.

**Why it matters for AI**:

Vector addition is used constantly in AI systems - for combining information, updating calculations, and much more. Computers can perform billions of these additions per second!

# Scalar Multiplication and Dot Product

Two fundamental operations: scaling a vector (making it bigger or smaller) and combining two vectors into a single number.

**Scalar Multiplication (Scaling)**:

Multiply each number in the vector by the same value:
3 · [1, 2] = [3×1, 3×2] = [3, 6]

Think of it as:

* Multiplying by 2 = doubling the vector (twice as long)
* Multiplying by 0.5 = halving the vector (half as long)
* Multiplying by -1 = flipping the direction

**Real-world example**:

If a recipe for 2 people uses [2 eggs, 1 cup flour], then for 6 people:
3 × [2, 1] = [6 eggs, 3 cups flour]

**Dot Product**:

Multiply matching components and add them all up:
[1, 2, 3] · [4, 5, 6] = 1×4 + 2×5 + 3×6 = 4 + 10 + 18 = 32

The result is just one number, not a vector!

**What the dot product tells us**:

* Large positive result = vectors point in similar directions
* Zero = vectors are perpendicular (at right angles)
* Large negative result = vectors point in opposite directions

**Real-world example**:

Imagine comparing movie preferences. Your ratings: [5, 1, 4] (Action, Romance, Comedy). Friend's ratings: [4, 2, 5].

Dot product = 5×4 + 1×2 + 4×5 = 20 + 2 + 20 = 42

A high number means similar taste!

**In Python with NumPy**:

```python
import numpy as np
v = np.array([1, 2, 3])
scaled = 3 * v  # [3, 6, 9]

u = np.array([1, 2, 3])
v = np.array([4, 5, 6])
dot_product = u @ v  # 32
```

**Why it matters for AI**:

The dot product is how AI systems measure similarity between things. Similar items have high dot products, different items have low ones. This is the foundation of recommendation systems, search engines, and language models!

# Vector Norms

How do you measure the "size" of a vector? That's what norms do - they give us a single number representing how big a vector is.

**L2 Norm (The "straight line" distance)**:

This is the most common way to measure vector length:
||v||₂ = √(v₁² + v₂² + ... + vₙ²)

For vector [3, 4]: √(3² + 4²) = √(9 + 16) = √25 = 5

This is the Pythagorean theorem! If you remember a²+b²=c² from school, that's exactly what this is.

**L1 Norm (The "city block" distance)**:

Just add up the absolute values:
||v||₁ = |v₁| + |v₂| + ... + |vₙ|

For vector [3, 4]: |3| + |4| = 7

Imagine walking in a city with a grid of streets - you can't walk diagonally, only along blocks. That's L1 distance!

**Normalization (Making length = 1)**:

Sometimes we only care about direction, not size. Dividing by the length gives a "unit vector":
unit_v = v / ||v||

For [3, 4]: [3/5, 4/5] = [0.6, 0.8] - same direction, but length is now 1!

**In Python with NumPy**:

```python
import numpy as np
v = np.array([3, 4])

l2 = np.linalg.norm(v)  # 5.0 (straight line distance)
l1 = np.linalg.norm(v, ord=1)  # 7.0 (city block distance)
unit_v = v / np.linalg.norm(v)  # [0.6, 0.8]
```

**Why it matters for AI**:

Norms help us compare vectors fairly regardless of their original size. When comparing document similarity or image features, we often normalize first so longer documents don't unfairly dominate.

# What is a Matrix?

A matrix is a grid of numbers arranged in rows and columns - like a spreadsheet!

**Everyday examples**:

* A spreadsheet of student grades (rows = students, columns = subjects)
* Pixels in an image (rows and columns of color values)
* A seating chart (rows and columns of seats)

**Notation**:

We describe a matrix by its dimensions: rows × columns

```markdown
M = [[1, 2, 3],
     [4, 5, 6]]
```

This is a 2×3 matrix (2 rows, 3 columns).

**Accessing elements**:

Use [row, column] to get a specific number:

* M[0, 0] = 1 (first row, first column)
* M[1, 2] = 6 (second row, third column)

Remember: Python counts from 0, not 1!

**In Python with NumPy**:

```python
import numpy as np
M = np.array([[1, 2, 3],
              [4, 5, 6]])

print(M.shape)  # (2, 3) - 2 rows, 3 columns
print(M[0, 1])  # 2 - row 0, column 1
print(M[1])     # [4, 5, 6] - entire row 1
print(M[:, 0])  # [1, 4] - entire column 0
```

**Special matrices**:

* Identity matrix: 1s on the diagonal, 0s elsewhere (like multiplying by 1)
* Zero matrix: all zeros

**Why matrices matter for AI**:

Matrices are perfect for organizing data. An image is a matrix of pixels. A dataset of 1000 examples with 10 features is a 1000×10 matrix. AI systems process these matrices to find patterns and make predictions.

# Matrix Operations

Just like vectors, we can add matrices and multiply them by numbers. Plus one new trick: the transpose!

**Matrix Addition**:

Add matching positions together (matrices must be same size):

```markdown
[[1, 2],   +   [[5, 6],   =   [[6, 8],
 [3, 4]]        [7, 8]]        [10, 12]]
```

**Scalar Multiplication**:

Multiply every number in the matrix:

```markdown
2 × [[1, 2],   =   [[2, 4],
     [3, 4]]        [6, 8]]
```

**Transpose (flip rows ↔ columns)**:

Rows become columns, columns become rows:

```markdown
Original:           Transposed:
[[1, 2, 3],    →    [[1, 4],
 [4, 5, 6]]          [2, 5],
                     [3, 6]]
```

A 2×3 matrix becomes a 3×2 matrix!

**In Python with NumPy**:

```python
import numpy as np
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

sum_matrix = A + B  # Add matching positions
scaled = 2 * A      # Multiply all by 2
transposed = A.T    # Flip rows and columns
```

**Why it matters**:

These basic operations are the building blocks for more complex calculations. Transpose is especially useful when we need to reorganize data or when dimensions don't quite line up for other operations.

# Matrix-Vector Multiplication

Matrix-vector multiplication is THE fundamental operation in neural networks. It transforms an input vector into an output vector.

**Formula**:

For matrix M (m×n) and vector v (n,):
(Mv)[i] = Σⱼ M[i,j] · v[j]

**Example**:

```markdown
[[1, 2],     [5]     [1·5 + 2·6]     [17]
 [3, 4]]  ×  [6]  =  [3·5 + 4·6]  =  [39]
```

**Shape rule**:

(m×n) @ (n,) = (m,)
The inner dimensions must match!

**Geometric meaning**:

Matrix multiplication is a linear transformation:

* Rotation, scaling, shearing, projection
* Each output is a weighted combination of inputs

**In NumPy**:

```python
M = np.array([[1, 2], [3, 4]])  # (2, 2)
v = np.array([5, 6])  # (2,)
result = M @ v  # (2,)
```

**In neural networks**:

Single neuron layer: y = Wx + b

* W: weight matrix (out_features × in_features)
* x: input vector (in_features,)
* b: bias vector (out_features,)
* y: output vector (out_features,)

This is what happens in every Linear/Dense layer!

# Matrix-Matrix Multiplication

Matrix-matrix multiplication composes transformations and processes batches of data simultaneously.

**Formula**:

For A (m×n) and B (n×p):
(AB)[i,j] = Σₖ A[i,k] · B[k,j]

**Shape rule**:

(m×n) @ (n×p) = (m×p)

Inner dimensions must match!

**Key insight**: NOT element-wise!

```markdown
A * B  # Element-wise (different!)
A @ B  # Matrix multiplication
```

**Example**:

```markdown
[[1, 2],     [[5, 6],     [[1·5+2·7, 1·6+2·8],     [[19, 22],
 [3, 4]]  @   [7, 8]]  =   [3·5+4·7, 3·6+4·8]]  =   [43, 50]]
```

**Geometric meaning**:

Composition of transformations:

* If A rotates and B scales, AB does both
* Order matters! AB ≠ BA (usually)

**In neural networks**:

* Batch processing: X @ W.T where X is (batch, features)
* Multi-layer: y = W₂(W₁x) = (W₂W₁)x
* Attention: Q @ K.T computes attention scores

# Linear Transformations: y = Wx + b

The equation y = Wx + b is the heart of neural networks. Every fully connected layer performs exactly this operation.

**Components**:

* W: weight matrix (out_features × in_features)
* x: input vector (in_features,)
* b: bias vector (out_features,)
* y: output vector (out_features,)

**Example**:

Input: 4 features → Output: 3 features

```python
W = np.random.randn(3, 4)  # (3, 4)
x = np.random.randn(4)     # (4,)
b = np.random.randn(3)     # (3,)
y = W @ x + b              # (3,)
```

**What it does**:

* W @ x: linear combination of inputs (weighted sum)
* b: shift the result (allows non-zero output when x=0)

**Why bias?**

Without bias, output is always 0 when input is 0.
Bias lets the neuron have a "default" activation.

**In PyTorch**:

```python
layer = nn.Linear(4, 3)  # in=4, out=3
y = layer(x)  # Does W @ x + b internally
```

Stacking these layers with non-linear activations creates deep networks!
