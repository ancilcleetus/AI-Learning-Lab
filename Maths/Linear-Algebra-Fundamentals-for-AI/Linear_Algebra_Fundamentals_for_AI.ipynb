# What is a Vector?

A vector is simply an ordered list of numbers. Think of it as a way to store multiple related values together.

**Everyday examples**:

* Your location on a map: [latitude, longitude] - that's a 2D vector!
* A color: [red, green, blue] like [255, 128, 0] for orange - a 3D vector
* A student's grades: [Math: 85, Science: 92, English: 78] - a 3D vector

**Mathematical notation**:

A vector is written as: v = [v₁, v₂, ..., vₙ]

Examples:

* 2D vector: [3, 4] (two numbers)
* 3D vector: [1, 2, 3] (three numbers)
* You can have vectors with hundreds or thousands of numbers!

**Picture it**:

A 2D vector [3, 4] is like an arrow:

* Starting point: the origin (0,0)
* Ending point: (3, 4)
* Direction: 3 units right, 4 units up
* Length: √(3² + 4²) = 5

**In Python with NumPy**:

```python
import numpy as np
vec = np.array([3, 4])  # Create a vector
length = np.linalg.norm(vec)  # Compute length = 5.0
```

**Why vectors matter for AI**:

Computers can't understand words or images directly. Instead, we convert everything into vectors of numbers. An image becomes a long list of pixel values. A word becomes a list of numbers that capture its meaning. This course will show you how!

## Exercises

### Import numpy as np and create a 2D vector `[3, 4]` as a NumPy array.

````python
import numpy as np

vec = np.array([3, 4])
print(vec)
````

### Compute the magnitude (length) of the vector using `np.linalg.norm()`.

````python
length = np.linalg.norm(vec)
print(length)
````

### Create a 3D vector `[1, 2, 3]` and compute its magnitude.

````python
vec_3d = np.array([1, 2, 3])
print(vec_3d)
length = np.linalg.norm(vec_3d)
print(length)
````

### Create a high-dimensional vector with 10 random values using `np.random.randn()` and compute its magnitude.

````python
vec_10d = np.random.randn(10)
print(vec_10d)
length = np.linalg.norm(vec_10d)
print(length)
````

### Print the shape of the high-dimensional vector to see its dimensions.

````python
print(vec_10d.shape)
````
