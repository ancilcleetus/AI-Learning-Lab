# NumPy: Creating Arrays

NumPy is the fundamental package for numerical computing in Python. It provides fast, memory-efficient arrays and mathematical operations. Deep learning frameworks like PyTorch are built on similar concepts.

**Why NumPy?**

* Python lists are slow for numerical operations
* NumPy arrays are 10-100x faster
* Optimized C code under the hood
* Foundation for scientific Python ecosystem

**Creating arrays**:

```python
import numpy as np

# From Python list
arr = np.array([1, 2, 3, 4, 5])

# Zeros and ones
zeros = np.zeros((3, 4))  # 3x4 array of zeros
ones = np.ones((2, 3))    # 2x3 array of ones

# Range of values
arange_arr = np.arange(0, 10, 2)  # [0, 2, 4, 6, 8]
linspace_arr = np.linspace(0, 1, 5)  # 5 values from 0 to 1

# Random arrays
random_arr = np.random.randn(3, 3)  # 3x3 random normal
```

**Array attributes**:

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)   # (2, 3) - dimensions
print(arr.dtype)   # dtype('int64') - data type
print(arr.size)    # 6 - total elements
print(arr.ndim)    # 2 - number of dimensions
```

**In deep learning**:

```python
# Image as array (height, width, channels)
image = np.zeros((224, 224, 3))

# Batch of images (batch, height, width, channels)
batch = np.random.randn(32, 224, 224, 3)

# Network weights
weights = np.random.randn(784, 128) / np.sqrt(784)
```

NumPy arrays are the foundation for understanding tensors in PyTorch and TensorFlow.
