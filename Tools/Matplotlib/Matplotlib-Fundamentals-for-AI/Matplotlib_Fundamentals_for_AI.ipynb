# Setup

Before we jump in, let's quickly run the setup cell. It checks the environment (GPU, library versions) and sets random seeds for reproducibility — we want identical results on every run.

````python
# 🔧 Setup: Run this cell first!
# Environment check + random seeds for reproducibility

import sys
import time
import random
import numpy as np
import pandas as pd
import matplotlib as mpl
import torch
import torch.nn as nn

print(f"📦 Python       {sys.version.split()[0]}")
print(f"🔢 NumPy        {np.__version__}")
print(f"🐼 Pandas       {pd.__version__}")
print(f"📊 Matplotlib   {mpl.__version__}")
print(f"🔥 PyTorch      {torch.__version__}")

# GPU check (only needed for optional deep-feature experiments)
if torch.cuda.is_available():
    device = torch.device("cuda")
    print(f"✅ GPU available: {torch.cuda.get_device_name(0)}")
    print(f"   Memory: {torch.cuda.get_device_properties(0).total_memory / 1e9:.1f} GB")
else:
    device = torch.device("cpu")
    print("⚠️ No GPU detected")

DEVICE_LABEL = "GPU" if device.type == "cuda" else "CPU"

# Random seeds for reproducibility
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

print(f"\n🎲 Random seed set to {SEED} (Python, NumPy, PyTorch)")

%matplotlib inline
%config InlineBackend.figure_format = 'jpeg'   # photos as JPEG, not PNG — ~10× smaller notebook
# Remove character limits for all columns of Pandas dataframes permanently in this session
pd.set_option('display.max_colwidth', None)
````

# Matplotlib Basics for Visualization

Matplotlib is essential for visualizing training progress and model behavior in deep learning.

**Basic line plot (for loss curves)**:

```python
import matplotlib.pyplot as plt

losses = [2.5, 2.0, 1.5, 1.2, 1.0]
plt.plot(losses)
plt.xlabel('Iteration')
plt.ylabel('Loss')
plt.title('Training Loss')
plt.show()
```

**Multiple lines**:

```python
train_loss = [2.5, 2.0, 1.5]
val_loss = [2.6, 2.2, 1.8]
plt.plot(train_loss, label='train')
plt.plot(val_loss, label='val')
plt.legend()
plt.show()
```

**Scatter plot (for embeddings)**:

```python
x = [1, 2, 3, 4]
y = [1, 4, 2, 3]
plt.scatter(x, y)
plt.show()
```

**Image display (for attention maps)**:

```python
import numpy as np
data = np.random.rand(10, 10)
plt.imshow(data, cmap='hot')
plt.colorbar()
plt.show()
```

**Subplots**:

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot([1, 2, 3])
axes[1].scatter([1, 2], [3, 4])
plt.show()
```

Visualization helps you understand:

* Training dynamics (is loss decreasing?)
* Model predictions vs actual values
* Learned representations (embeddings)
* Attention patterns in transformers

## Exercises

### Import matplotlib.pyplot as plt and create a line plot of `[3.0, 2.5, 2.0, 1.5, 1.0]`.

````python
import matplotlib.pyplot as plt

losses = [3.0, 2.5, 2.0, 1.5, 1.0]
plt.plot(losses)
plt.show()
````

### Add `xlabel='Step'` and `ylabel='Loss'` to the plot.

````python
plt.plot(losses)
plt.xlabel('Step')
plt.ylabel('Loss')
plt.show()
````

### Plot two lines: `train_loss=[2.5, 2.0, 1.5]` and `val_loss=[2.6, 2.2, 1.8]` with labels, then add a legend.

````python
train_loss = [2.5, 2.0, 1.5]
val_loss = [2.6, 2.2, 1.8]
plt.plot(train_loss, label='train')
plt.plot(val_loss, label='val')
plt.legend()
plt.show()
````

### Create a scatter plot with `x=[1, 2, 3, 4]` and `y=[2, 4, 1, 3]`.

````python
x, y = [1, 2, 3, 4], [2, 4, 1, 3]
plt.scatter(x, y)
plt.show()
````

### Import numpy, create a 5x5 random matrix, display it with imshow using `cmap='viridis'`, and add a colorbar.

````python
import numpy as np

data = np.random.rand(5, 5)
plt.imshow(data, cmap='viridis')
plt.colorbar()
plt.show()
````
