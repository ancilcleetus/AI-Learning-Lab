# A neuron is embarrassingly simple

A neural network looks intimidating from the outside, but the unit it is made of fits on one line: multiply each input by a weight, add them up, add a bias.

Everything else in this course — CNNs, RNNs, transformers — is this same operation repeated with more structure. Learn it once here and the rest is arrangement.

## Compute one neuron by hand

The next cell holds three inputs, three weights and one bias. The line `(x * w).sum() + b` is the entire neuron: an elementwise multiply, a sum, and a shift.

Run it. The value it prints, z, is called the pre-activation — the neuron's raw opinion before any non-linearity touches it.

```python
x = torch.tensor([0.5, -1.2, 2.0])  # 3 inputs
w = torch.tensor([0.8, 0.3, -0.5])  # 3 weights
b = torch.tensor(0.1)  # 1 bias

z = (x * w).sum() + b
# z = torch.dot(x, w) + b
print(f"Pre-activation, z = {z.item():.4f}")
```

## The bias moves the answer alone

Weights decide how much each input matters. The bias decides where the neuron sits when every input is zero — it slides the whole output up or down.

```python
BIAS = 1.0  # Increase bias by 0.9 (0.1 + 0.9 = 1.0)
b = torch.tensor(BIAS)
z = (x * w).sum() + b
print(f"Pre-activation, z = {z.item():.4f}")
```

Now z moves by exactly +0.9 $(0.0400 - -0.8600 = 0.9)$ compared to the previous run.

## The non-linearity that makes it deep

Stack a thousand of these linear neurons and you still only get a straight line. ReLU fixes that: it keeps positive values and clamps negatives to zero.

```python
pre_activations = torch.tensor([-0.9, 0.4, -0.2, 1.6, 0.0, 2.3, -1.1, 0.7])
activations = torch.relu(pre_activations)
print(f"Activations: {activations}")
```
