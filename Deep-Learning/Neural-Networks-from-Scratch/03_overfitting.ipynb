# Your model can score 100% and still be useless

In the last lesson accuracy climbed and you called it learning. But you measured on the exact same data the model trained on — the equivalent of grading an exam with the answer key taped to the desk.

Overfitting is when a network memorises the training set instead of learning the pattern behind it. This lesson gives you the two tools every practitioner uses to catch it: a **validation split** and **dropout**.

## Step 1 — hold data back

`random_split` carves the dataset into a training portion and a validation portion. The model never sees the validation half during backprop, so its score there is an honest estimate of real-world performance.
