# Introduction

LLMs never see raw text, only tokens. This lecture shows how a tokenizer splits text with Byte Pair Encoding (BPE), and why that drives cost.

* Why words split in strange places
* Why Hindi and other languages use more tokens
* Why ChatGPT, Claude and DeepSeek tokenize differently

**The entire lecture in one sentence**:

A tokenizer, trained with Byte Pair Encoding on huge amounts of text, turns your prompt into token IDs:

* Common text gets fewer, longer tokens
* Rare text gets more, shorter ones
* You pay for every token

![rainbow](https://raw.githubusercontent.com/ancilcleetus/AI-Learning-Lab/main/assets/rainbow-divider.png)

# What is a token?

LLMs don't read raw text. Before the model sees anything, a **tokenizer** splits your text into pieces called **tokens** and turns each piece into a number, its **token ID**. The model only ever works with those numbers.

<img src="https://raw.githubusercontent.com/ancilcleetus/AI-Learning-Lab/refs/heads/main/Generative-AI/AI-Engineering-Course-by-Sameer-Bhardwaj/Module-01-Foundations/01_Tokens/data/images/01_Tokens-01.png" width="800">

That's why tokens are called the **currency of LLMs**: context limits, speed and price are all measured in tokens, not words or characters.
