# Problem Statement

* **The task:** Predict the price of a house in Bengaluru given its characteristics — location, size (BHK), total square footage, number of bathrooms, etc. This is a **supervised regression** problem: the target `price` is a continuous number, measured in lakhs of rupees (₹ lakh).
* **The dataset:** [Bengaluru House Price Dataset from Kaggle](https://www.kaggle.com/datasets/amitabhajoy/bengaluru-house-price-data/) — ~13,300 real listings scraped from property websites, 9 columns.
* **Why this dataset:** It's realistically messy — mixed formats in `total_sqft` (ranges like "1133 - 1384", odd units like "34.46Sq. Meter"), dirty text in `size`, missing values, and extreme outliers. So the project exercises the full pipeline: cleaning → EDA → feature engineering → model → evaluation.
* **Success criteria:** Build a Linear Regression model and evaluate it with MAE, RMSE, and R² on a held-out test set.
