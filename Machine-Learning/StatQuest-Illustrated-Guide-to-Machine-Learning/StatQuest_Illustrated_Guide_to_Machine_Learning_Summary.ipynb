# Fundamental Concepts in Machine Learning

* Machine Learning => Collection of tools and techniques that transform data into (hopefully good) decisions by making
    * classifications (eg: whether or not someone will love a movie) => we call it **Classification**
    * quantitative predictions (eg: how tall someone is) => we call it **Regression**
* Classification => Example: Classification Tree to predict whether a person likes StatQuest or not
* Regression => Example: Black fit line (or) Green wiggle (or) Fancy method to predict height of a person given his weight.
    * Height => Dependent variable
    * Weight => Independent variable (or) Feature
* How to choose a machine learning method?
    * Performance (minimum error) on testing data
    * Even though Green Wiggle was fitting training data better than the Black fit line, Black fit line was chosen as the best method since the error on testing data was better for Black fit line compared to both the Green Wiggle & the Fancy method.
    * Just because an ML method fits training data well, it doesn't mean it will perform well on the testing data.
    * When an ML method fits training data really well, but makes poor predictions, we say that it is **Overfit** to the training data.

# Cross Validation

* **How do we pick the best points for training & the best points for testing?** => Cross Validation (the technique to select which data points should be used for training and which points should be used for testing)
* In Cross Validation, we do not worry whether we have selected the best points for training & the best points for testing. Instead, we use all of the data points for both training & testing.
* Given a dataset, we do the below steps for Cross Validation:
    * Randomize the dataset to change the order of data points
    * Do below steps in multiple iterations:
        * Split the dataset into equal sized blocks called folds
            * Dataset is large => Split into multiple folds if we have a big dataset => **K-fold Cross Validation** => No of iterations = K
            * Dataset is very small => Keep all except one point for training, one point for testing => **Leave-One-Out Cross Validation** => No of iterations = No of data points in the dataset
        * We take K-1 folds for training & 1 fold for testing where K can be no of folds or total no of data points (based on size of dataset)
        * We have different training data & different testing data in each iteration
            * Since each iteration uses different training data, it results in a **different fitted line**
            * A different fitted line with a different testing data results in **different prediction errors** for all the iterations
        * We calculate the error for each iteration
    * We can **average these errors** to get a general sense of how well our ML model will perform with future data
    * We can also compare these K errors to K errors made by another method.
* When the results are mixed (i.e. for some iterations, ML method 1 fairs well while for some other iterations, ML method 2 fairs well), how do we decide which method is better?
    * One way is to use **Statistics**

**Note:**

1. Reusing the same data for traning & testing => **Data Leakage**
2. When dataset is large, **10-Fold Cross Validation** is the recommended Cross Validation technique.

# Fundamental Concepts in Statistics

## Statistics

* Statistics provides us with **a set of tools to quantify the variation** that we find in everyday things.
* For Machine Learning, Statistics **helps us make predictions** and **quantify how confident we should be in those predictions**.
* **Example: No of French Fries we get everyday**
    * No of French Fries we get everyday varies
    * If we take note of no of french fries we get for the past few days,
        * Statistics can help us predict how many french fries we get today.
        * Statistics can also tell us how confident we should be in that prediction.
* **Example: A new medicine that helps some people but hurts some other people**
    * If we predict that the medicine can help, but we are not very confident in that prediction, the doctor might not recommend the medicine and can use a different therapy to help the patient.
* The first step in making predictions is to **identify the trend in the data** that we have collected and we can do that with **Histogram**.

## Histograms

* How to create a Histogram?
    * Lets say, we have collected height measurements from so many people and we want to identify the trend in this data.
    * We first **put the measurements in order**.
    * Then, we **divide the measurements into multiple bins** of equal size.
    * Then, we **stack measurements** falling in the same bins.
    * Then, we end up with a Histogram.
    * If the width of the bins are too high or too low, they are of not much help. The **width of the bins should not be too high or too low**.
* We can use the Histogram to estimate the probability of future measurements.
    * In our Histogram above, we see that **most measurements are close to the average height**.
    * Because most measurements are close to the average height, the **probability that next measurement is close to average height is high**.
    * **Probability that next measurement is short or tall is low** since shorter or taller measurements are rarer.
    * The **confidence we have in these measurements depends on the no of measurements**. The more measurements we have, the more confidence we can have in the estimate.
* Calculating Probabilities step-by-step with Histograms:
    * $\text{Probability that next measurement falls under box containing 3 bins around the average height} = \frac {\text{no of measurements in the box}}{\text{total no of measurements}} = \frac {12}{19} = 0.63 \Rightarrow $ 63% chance that the next measurement will fall within this box.
    * $\text{Probability that next measurement falls under box containing last bin (probability of tall)} = \frac {\text{no of measurements in the box}}{\text{total no of measurements}} = \frac {1}{19} = 0.05 \Rightarrow $ 5% chance that the next measurement will fall within this box.
    * $\text{Probability that next measurement falls under box containing all the bins} = \frac {\text{no of measurements in the box}}{\text{total no of measurements}} = \frac {19}{19} = 1 \Rightarrow $ 100% chance that the next measurement will fall within this box; Maximum probability is 1.
    * $\text{Probability that next measurement falls under box containing a bin with zero measurements} = \frac {\text{no of measurements in the box}}{\text{total no of measurements}} = \frac {0}{19} = 0 \Rightarrow $ 0% chance that the next measurement will fall within this box; Minimum probability is 0.
        * The only reason that this bin is empty may be because we didn't take enough measurements.
        * If we take more measurements,
            * we either get more measurements in this bin
            * we become more confident that it should be empty
* Getting more measurements can be expensive or take a lot of time or both $\Rightarrow$ we can solve this problem with a **Probability Distribution**.
