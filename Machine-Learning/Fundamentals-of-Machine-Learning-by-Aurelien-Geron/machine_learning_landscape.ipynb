# What is Machine Learning?

- *Machine learning is the field of study that gives computers the ability to learn without being explicitly programmed. (Arthur Samuel, 1959)*

- *A computer program is said to learn from experience E with respect to some task T and some performance measure P, if its performance on T, as measured by P, improves with experience E. (Tom Mitchell, 1997)*

# Types of Machine Learning Systems

ML Systems are categorized based on:
* Learning Mode (how they are guided during training) => Supervised learning, Unsupervised learning, Semi-supervised learning, Self-supervised learning, Reinforcement learning
* Training Mode => Ability for learning incrementally => Batch vs Online learning
* Modeling Mode => By comparing new data point with known data points or by building a predictive model based on detected patterns in training data) => Instance-based vs Model-based learning

### Self-supervised learning

* Self-supervised learning involves
    * generation of a fully labeled dataset from a fully unlabeled dataset
    * training of a model on the above dataset
    * tweaking of the model to serve the particular ML task
    * use of a supervised learning algorithm for fine-tuning in the final step
* Eg: Pet Image Classification task
    * Train Pet Image Repairing Model with
        * inputs to the model => Unlabeled pet images with mask
        * labels => Original pet images
    * Once it is performing well (repairs pet images correctly), tweak model architecture to predit pet species instead of repairing images
    * Fine-tune model on labeled dataset (inputs => Original pet images, labels => pet species)
* Self-supervised learning tasks include Classification, Regression etc.

## Training Mode (Batch vs Online learning)

### Batch learning

* Model trained using all available data, typically offline => Offline learning
* Deployed in production without learning anymore from new data
* Data drift or Model decay => Needs to retrain regularly on up-to-date data (train from scratch on old data + new data)
* Training on full dataset is slow & resource-heavy => costly & slow adaptation to change
* Eg: Random Forests => can only be trained from scratch on the full dataset

### Online learning

* Model trained incrementally on new data using mini-batches online => Online learning
* Each learning step is fast & resource-light => cheap & fast adaptation to change
* Eg: Gradient descent algorithm

## Modeling Mode (Instance-based vs Model-based Learning)

### Instance-based learning

* Prediction based on a *measure of similarity* with known instances
* Need to deploy a copy of training set to production => best for small dataset models; not recommended for models with big datasets

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/machine_learning_landscape_16.png", width=500))
display(Markdown("*Instance-based learning: in this example we consider the class of the three nearest neighbors in the training set*"))
```

### Model-based learning

* Instead of predicting from a set of instances, a suitable model is selected to capture the relationships in the training data
* After fitting this model on training data, it is used to make predictions on new data

# Main Challenges of Machine Learning

Three things that can go wrong in ML are:
* bad data
* bad model
* bad deployment

## Bad data

* Insufficient quantity of training data
* Nonrepresentative training data
* Poor-quality data
* Irrelevant features
    * Solution: Feature engineering which involves
        * Feature selection => selection of most useful features
        * Feature extraction => combining existing features to create more useful features
        * Creation of new features from new data

## Bad model

* Overfitting the training data
    * Occurs when model is too complex relative to amount and noisiness of training data
    * Solution:
        * Simplify model
            * by reducing no of parameters
            * by reducing no of features in training data
            * by constraining the model with regularization (eg: set some parameters to zero)
        * Gather more training data
        * Reduce noise in training data (e.g., fix data errors and remove outliers)

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/machine_learning_landscape_18.png", width=500))
display(Markdown("*Overfitting the training data*"))
```

```python
from IPython.display import Image, display, Markdown
display(Image(filename="data/images/machine_learning_landscape_19.png", width=500))
display(Markdown("*Dotted line => Linear model on partial data (circles)*"))
display(Markdown("*Solid line => Linear model on full data (circles & squares)*"))
display(Markdown("*Dashed line => Regularized linear model on partial data (circles)*"))
display(Markdown("*Regularization forced the model to have a smaller slope and thus reduced the risk of overfitting*"))
```

* Underfitting the training data
    * Occurs when model is too simple to learn the underlying structure of the training data
    * Solution:
        * Select a more powerful model, with more parameters
        * Feed better features to the learning algorithm (feature engineering)
        * Reduce the constraints on the model (for example by reducing the regularization hyperparameter)

## Bad deployment

* Model too complex to maintain
* Model is slow (large inference time)
* Model takes large memory
* Model may not scale properly
* Model not updated often

## Data Mismatch

* Training set may not be exactly representative of the data the model sees in production
* Validation & test sets must be as representative as possible of the data in production
* Eg: Flower Species Classifier Mobile App
    * Training set => Flower images from web
    * Validation & Test sets => Flower images captured using mobile app
    * Performance in Training set is good & that of Validation set is poor => Either due to overfitting of train data or data mismatch between web images & mobile images
        * Solution:
            * Keep some web images as Train-Dev set
            * Performance in Train-Dev set is poor => Overfitting => simplify model, get more training data, reduce noise in training data (e.g., fix data errors and remove outliers)
            * Performance in Train-Dev set is good => Data mismatch => preprocess web images to make them look more like the pictures taken by the mobile app & retrain the model
            * Once Performance in Train-Dev & Dev sets are good, evaluate in Test set
